Los Tipos de Datos Abstractos (TDA) permiten representar modelos de la vida real dentro de un entorno virtual, definiendo los atributos que describen sus características y las operaciones que pueden realizarse sobre ellos, como insertar, eliminar o buscar elementos. Este proceso de abstracción convierte una situación cotidiana —como una pila de ropa o una cola en el supermercado— en un modelo computacional manipulable.
En ciencias de la computación, los TDA son la base para construir estructuras de datos como pilas, colas, listas, árboles o grafos, cada una con su propio comportamiento y complejidad. Estas pueden implementarse con variables estáticas (vectores o matrices) o dinámicas (punteros), según las necesidades del modelo. Los TDA proporcionan así una forma estructurada de organizar y operar sobre los datos de manera eficiente y coherente.

Para poder definir la estructura interna de un TDA y sus funciones asociadas previamente es nece-
sario que se definan algunos conceptos básicos:
1. Variable dinámica: una variable dinámica es aquella que se puede crear y destruir en tiempo
de ejecución a diferencia de una variable estática que lo hace en tiempo de compilación. Este
tipo de variables son administradas mediante el uso de punteros.

2.   Puntero: el puntero es un tipo de dato que se utiliza para almacenar una dirección de memoria
de una variable dinámica –es decir se podría decir que un puntero apunta a una variable diná-
mica­ – esto también se conoce como dirección de referencia.
3. Nodo: es un tipo de datos que se define mediante el uso de una clase –dado que en Python no
existen los registros– para representar a cada elemento de nuestra estructura de datos. De-
pendiendo de esta pueden variar sus atributos, pero en su esencia un nodo debe tener míni-
mamente un campo donde se almacene la información y otro que lo enlace con el siguiente o
anterior nodo de la estructura.

4. Variables huérfanas: cuando una variable dinámica deja de ser apuntada por un puntero, que-
dará en memoria pero no podrá accederse a ella porque su dirección se perdió. En Python
cuando ocurre esto el intérprete elimina automáticamente dicha variable para liberar espacio
en memoria. Por lo que para eliminar una variable dinámica solo basta con cambiar la direc-
ción del puntero que la señala.

```python
# Ejemplo: variables estáticas (inmutables) vs dinámicas (mutables)
# Variables estáticas (números: inmutables)
num1 = 1
num2 = num1
# num2 recibe un nuevo entero; num1 no cambia porque los int son inmutables
num2 = num2 + 3
print("valor num1", num1, "valor num2", num2)
print("id num1:", id(num1), "id num2:", id(num2))
print()

# Variables dinámicas (listas: mutables)
lista1 = [1, 2, 3]
lista2 = lista1  # alias: ambas variables referencian la misma lista
# Modificar a través de la segunda referencia afecta a la primera
lista2[1] = 7
lista2.append(87)
print("valor lista1", lista1, "valor lista2", lista2)
print("id lista1:", id(lista1), "id lista2:", id(lista2))
print()

# Para crear una copia independiente (no alias), usar copy() o lista[:]
lista3 = lista1.copy()
lista3[0] = 99
print("lista1 (después)", lista1)
print("lista3 (copia modificada)", lista3)
print("id lista3:", id(lista3))
```

Una ventaja de trabajar con punteros es que almacenan una dirección de memoria por lo que se
puede tener más de un puntero que apunten a una misma variable dinámica. Las operaciones útiles
de comparaciones que se puede hacer entre punteros son dos, si ambas variables son iguales –es
decir si están apuntando a la misma variable dinámica– o si los punteros son distintos –si no están
apuntando a la misma variable dinámica­ –. Esto se puede ver en el siguiente código, la salida del programa
será para el primer caso “Apuntan a diferentes variable dinámica” y “Apuntan a la misma variable
dinámica” para el segundo caso. El único valor conocido que el programador le puede asignar a un
puntero es vacío o None (en el caso de Python) –dado que no es posible conocer las direcciones de
memoria–. Esto significa que el puntero no apunta a ninguna variable, también se puede asignar
una dirección de memoria que este almacenada en otro puntero. Las demás direcciones son asigna-
das automáticamente por el sistema operativo cuando creamos las variables dinámicas.

```python
# Comparaciones: igualdad (==) vs identidad (is) y ejemplo de aliasing
puntero1 = [0, 1]
puntero2 = [0, 1]  # dos listas con mismo contenido pero objetos distintos
print('puntero1:', puntero1)
print('puntero2:', puntero2)
print('puntero1 == puntero2 ->', puntero1 == puntero2)  # compara contenido
print('puntero1 is puntero2 ->', puntero1 is puntero2)  # compara identidad (misma instancia)
print('id puntero1:', id(puntero1), 'id puntero2:', id(puntero2))
if puntero1 == puntero2:
    print('== : Tienen el mismo contenido (no necesariamente la misma instancia)')
else:
    print('== : Contenido distinto')
if puntero1 is puntero2:
    print('is : Apuntan a la misma variable dinámica (misma instancia)')
else:
    print('is : Apuntan a diferentes variables dinámicas (instancias distintas)')

# Ahora reasignamos para crear alias (misma referencia)
puntero2 = {}
puntero1 = puntero2  # ahora ambas variables referencian el mismo diccionario
print('\nDespués de puntero2 = {} y puntero1 = puntero2:')
print('puntero1:', puntero1)
print('puntero2:', puntero2)
print('puntero1 == puntero2 ->', puntero1 == puntero2)
print('puntero1 is puntero2 ->', puntero1 is puntero2)
print('id puntero1:', id(puntero1), 'id puntero2:', id(puntero2))
if puntero1 is puntero2:
    print('Apuntan a la misma variable dinámica')
else:
    print('Apuntan a diferentes variable dinámica')
```

Un nodo es un tipo de datos que se define con una clase como se observa en el siguiente código, con el cual
se representan cada uno de los elementos de nuestra colección o estructura de datos. A diferencia
de las estructuras estáticas en las que se accede a los elementos mediante índices o posiciones como
en el caso de los vectores o las matrices. Las estructuras dinámicas son un conjunto de nodos enla-
zados entre sí –mediante el uso de punteros–, de manera tal que a partir de un determinado nodo
se puede acceder al anterior, al siguiente o a ambos, dependiendo de la estructura del nodo definida
y de la naturaleza de la estructura que se quiera modelar.

Las estructuras dinámicas se diferencian de las estáticas porque permiten agregar o eliminar elementos durante la ejecución del programa, optimizando así el uso de memoria al ocupar solo el espacio necesario. En cambio, las estructuras estáticas definen su tamaño en tiempo de compilación, lo que puede generar desperdicio de memoria o limitar la flexibilidad para modificar su contenido.
Un nodo enlazado, como muestra la figura, está formado por dos campos: uno que almacena la información y otro que actúa como enlace o puntero hacia el siguiente nodo. Esta composición permite crear cadenas de nodos conectados entre sí, base fundamental de estructuras dinámicas como listas, pilas y colas.

El nodo simplemente enlazado es un tipo de dato recursivo, ya que contiene un campo con la información y otro que apunta a otro nodo del mismo tipo, formando así una estructura que se define a sí misma. Esta característica permite recorrer la lista de forma secuencial hasta llegar al último nodo, identificado cuando el enlace siguiente es None.
En el ejemplo de la figura, se muestra cómo crear dinámicamente una lista de palabras usando nodos enlazados. Cada palabra ingresada por el usuario genera un nuevo nodo conectado al anterior, y mediante un puntero se puede recorrer toda la lista para imprimir su contenido. Este proceso ilustra el manejo dinámico y recursivo de estructuras enlazadas en memoria.

```python
class Nodo:
    """Clase Nodo simplemente enlazado."""
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None  # referencia al siguiente nodo (inicialmente None)

# Cargar 10 palabras iniciales en nodos (sin entrada interactiva)
palabras_iniciales = ['uno', 'dos', 'tres', 'cuatro', 'cinco', 'seis', 'siete', 'ocho', 'nueve', 'diez']
# Crear la lista enlazada a partir de la lista de palabras
if palabras_iniciales:
    auxiliar = Nodo(palabras_iniciales[0])  # cabeza de la lista
    nodo_auxiliar = auxiliar
    for palabra in palabras_iniciales[1:]:
        nodo = Nodo(palabra)
        nodo_auxiliar.siguiente = nodo
        nodo_auxiliar = nodo
else:
    auxiliar = Nodo('Primer nodo')

# Recorrer e imprimir la lista creada
auxiliar_recorre = auxiliar
while auxiliar_recorre is not None:
    print(auxiliar_recorre.dato)
    auxiliar_recorre = auxiliar_recorre.siguiente
```

```python
class Nodo(object):
    """Clase Nodo simplemente enlazado."""
    def __init__(self, info=None, siguiente=None):
        self.info = info
        self.sig = siguiente
    def __repr__(self):
        return f"Nodo({self.info})"

# --- Ejemplo de uso ---
# Crear nodos con valores (como en la figura)
nodo1 = Nodo(19)       # variable estática (valor) representado en un nodo
nodo2 = Nodo(345)      # variable dinámica apuntada por un puntero
nodo3 = Nodo(124)      # otra variable dinámica (huérfana si nadie la referencia)

# Enlazar: n1 -> n2 -> n3
nodo1.sig = nodo2
nodo2.sig = nodo3

print('Recorrido desde nodo1:')
"""cur"""
cur = nodo1 # iniciar en la cabeza la variable cur es un puntero a un nodo (nodo1)
while cur is not None: # recorrer hasta el final (None)
    print('  ', cur, '-> sig.info =', getattr(cur.sig, 'info', None)) # mostrar info y la del siguiente
    cur = cur.sig   # avanzar al siguiente

# Mostrar ids para visualizar referencias (direcciones simuladas)
print('\nIDs:')
print(' id n1:', id(nodo1))
print(' id n2:', id(nodo2))
print(' id n3:', id(nodo3))

# Ejemplo: hacer que n2 quede huérfano (desvincular desde su puntero)
# Si eliminamos la referencia desde n1, n2 seguirá existiendo mientras la variable n2 exista.
nodo1.sig = None
print('\nDespués de n1.sig = None:')
print(' n1.sig ->', nodo1.sig)
print(' n2 ->', nodo2, 'n2.sig ->', nodo2.sig)

# Para simular que n2 queda verdaderamente huérfano eliminamos la variable n2
# (si no existen otras referencias, el recolector liberaría el objeto)
temp_id = id(nodo2)
del nodo2 # eliminar la variable que referenciaba el nodo2
import gc # importar el módulo de recolección de basura
gc.collect() # forzar recolección de basura
print('\nSe eliminó la variable n2; si no hay más referencias, el objeto con id', temp_id, 'queda listo para ser recolectado por GC.')

# Nota: en este notebook no podemos mostrar directamente la liberación, pero
print('Nota: si quieres, podemos generar un diagrama visual (PNG) que muestre las referencias.')
```
